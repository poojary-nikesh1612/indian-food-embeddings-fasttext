```python
import pandas as pd
```

```python
data=pd.read_csv('Cleaned_Indian_Food_Dataset.csv')
data.head()
```

```python
text=data.TranslatedInstructions[0]
print(text)
```

```python
import re

def preprocess(text):
  text=re.sub(r"[^\w\s]"," ",text)
  text=re.sub(r"\s+"," ",text).strip()
  return text.lower()

print(preprocess(text))
```

```python
data['TranslatedInstructions']=data['TranslatedInstructions'].map(preprocess)
data['TranslatedInstructions'][0]
```

```python
data.to_csv('indian_recipes.txt',columns=['TranslatedInstructions'],header=None,index=False)
```

# Skip Gram

```python
import fasttext

model=fasttext.train_unsupervised('indian_recipes.txt')
```

```python
model.get_nearest_neighbors('dosa')
```

```python
model.get_analogies('neer','dosa','south indian')
```
